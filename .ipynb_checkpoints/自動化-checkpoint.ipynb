```python
import pandas as pd
import os
from datetime import datetime
```

```python
#保存先フォルダ
folder_path = "sales"
os.makedirs(folder_path, exist_ok=True)
```

```python
#今日の日付を使ってファイル名を作成
today = datetime.now().strftime("%Y%m%d")
file_name = f'{today}.csv'
file_path = os.path.join(folder_path, file_name)
```

```python
#サンプルデータを作成
data = {
    "日付": [datetime.now().strftime('%Y-%m-%d')] * 4,
    '店舗名': ['東京店', '大阪店', '東京店', '大阪店'],
    '商品名': ['コーヒー', 'サンドイッチ', '紅茶', 'ケーキ'],
    '数量': [3, 2, 5, 1],
    '単価': [300, 500, 250, 400]
}
```

```python
#DataFrameに変換
df = pd.DataFrame(data)
```

```python
#CSVファイルとして保存
df.to_csv(file_path, index=False, encoding='utf-8-sig')
print(f'{file_path}に売り上げデータを書き出しました。')
```
